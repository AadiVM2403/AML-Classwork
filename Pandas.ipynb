```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset('titanic')

print("=== 1. Average Age of Survivors vs Non-Survivors ===")
avg_age_survived = df.groupby('survived')['age'].mean().rename(index={0: 'Non-Survivor', 1: 'Survivor'})
print(avg_age_survived.to_string())
print("\n" + "="*50 + "\n")
```

```python
print("=== 2. Passengers Older Than 60 Years ===")
# Adjust column names based on dataset version ('age', 'sex', 'class', etc.)
older_than_60 = df[df['age'] > 60][['sex', 'age', 'class', 'fare', 'survived']]
print(f"Total passengers > 60 years old: {len(older_than_60)}\n")
print(older_than_60.head(10))  # Showing first 10 rows
print("\n" + "="*50 + "\n")
```

```python
print("=== 3. Average Fare by Passenger Class ===")
avg_fare_class = df.groupby('pclass')['fare'].mean()
print(avg_fare_class.to_string())
```
