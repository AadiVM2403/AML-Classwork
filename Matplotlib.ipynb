```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame= True)

df = iris.frame

df.head()
```

```python
plt.scatter(df['sepal length (cm)'],
            df['petal length (cm)'])

plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```

```python
plt.hist(df['petal width (cm)'])

plt.title("Distribution")
plt.show()
```

```python
species = ['Setosa','Versicolor','virginica']

count=df['target'].value_counts()

plt.bar(species,count)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd


days = list(range(1, 32))


bangalore_temp = [
    23, 24, 24, 23, 24, 24, 24, 24, 25, 24,
    25, 25, 26, 26, 25, 25, 23, 23, 24, 23,
    24, 22, 24, 24, 23, 24, 25, 24, 24, 24, 24
]

hyderabad_temp = [
    26, 26, 25, 24, 26, 27, 27, 28, 28, 29,
    29, 29, 30, 29, 29, 28, 25, 26, 26, 26,
    24, 24, 26, 26, 26, 26, 25, 26, 24, 24, 26
]


df = pd.DataFrame({
    'Day': days,
    'Bangalore': bangalore_temp,
    'Hyderabad': hyderabad_temp
})


plt.figure(figsize=(14, 6))
plt.plot(df['Day'], df['Bangalore'], marker='o', linewidth=2, label='Bangalore (July 2026)', color='#0066cc')
plt.plot(df['Day'], df['Hyderabad'], marker='s', linewidth=2, label='Hyderabad (July 2026)', color='#ff6600')

plt.title('Actual Daily Temperatures in July 2026: Bangalore vs Hyderabad', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Day of July', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.xticks(range(1, 32))
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()

```
